```python
rows_to_drop = []



for index, row in data_4.iterrows():

    if pd.isnull(row['reviews.doRecommend']):

        if row['reviews.rating'] > 3:

            data_4.at[index, 'reviews.doRecommend'] = True

        elif row['reviews.rating'] < 3:

            data_4.at[index, 'reviews.doRecommend'] = False

        elif row['reviews.rating'] == 3:

            rows_to_drop.append(index)

    if pd.isnull(row['reviews.doRecommend']) and pd.isnull(row['reviews.rating']):

        rows_to_drop.append(index)



data_4.drop(rows_to_drop, inplace=True)



data_4
```

```python
from nltk.stem import PorterStemmer

from nltk.corpus import stopwords

from nltk.tokenize import word_tokenize

import re



ps = PorterStemmer()

stop_words = set(stopwords.words("english"))



reviews = []

recommendations = []

for i in range(len(data_4)):

    title = data_4['reviews.title'].iloc[i]

    text = data_4['reviews.text'].iloc[i]

    recommend = data_4['reviews.doRecommend'].iloc[i]



    if not isinstance(title, str):

        title = ""

    if not isinstance(text, str):

        text = ""



    combined_text = f"{title} {text}"



    keep_alphabet_only = re.sub('[^a-zA-Z]', ' ', combined_text)  # Remove non-alphabetical characters

    lowrecase_text = keep_alphabet_only.lower()  # Convert to lowercase

    tokenized_text = word_tokenize(lowrecase_text)  # Tokenize the text

    remove_stopwords = [ps.stem(word) for word in tokenized_text if word not in stop_words]  # Remove stopwords and stem words

    cleaned_text = ' '.join(remove_stopwords)  # Join tokens back into a single string



    reviews.append(cleaned_text)

    recommendations.append(int(recommend))

reviews
```

```python
cv = CountVectorizer()

x_text = cv.fit_transform(reviews)
```

```python
x_recommend = np.array(recommendations).reshape(-1, 1)

x = hstack([x_text, x_recommend])
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size = 0.15, random_state = 2)
```

```python
import google.generativeai as genai
import numpy as np
from scipy.sparse import hstack
import re
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
import nltk

# Download required NLTK data
nltk.download('punkt')
nltk.download('stopwords')

# Initialize components
ps = PorterStemmer()
stop_words = set(stopwords.words('english'))

# Configure Gemini API
API_KEY = "AIzaSyAGSUbpPI1q-U8sCJ9BVm_s78sR_F9XJxQ"  # Store securely
genai.configure(api_key=API_KEY)

def translate_urdu_to_english(text):
    """
    Translate Urdu or Roman Urdu text to English using Gemini API.
    """
    try:
        model = genai.GenerativeModel("gemini-pro")
        response = model.generate_content(f"Translate this Urdu text to English: {text}")
        
        if response and hasattr(response, 'text'):
            return response.text.strip()
        else:
            return "Translation failed: No response from API."
    except Exception as e:
        return f"Error: {str(e)}"

def preprocess_text(title, text):
    """
    Preprocess the text data for sentiment analysis.
    """
    combined_input = f"{title} {text}"
    clean = re.sub('[^a-zA-Z]', ' ', combined_input)
    clean = clean.lower()
    clean = word_tokenize(clean)
    clean = [ps.stem(word) for word in clean if word not in stop_words]
    return ' '.join(clean)

def analyze_review(cv, xgb, title, text, recommend, is_urdu=False):
    """
    Analyze a review using the trained model.
    """
    # Translate if the text is in Urdu
    if is_urdu:
        title = translate_urdu_to_english(title)
        text = translate_urdu_to_english(text)
        
        # Check for translation errors
        if "Error:" in title or "Error:" in text:
            return "Translation error occurred"
    
    # Preprocess the text
    clean_text = preprocess_text(title, text)
    
    # Transform the text using CountVectorizer
    new_review_text = cv.transform([clean_text])
    
    # Convert recommendation to binary
    recommend_binary = 1 if recommend.lower() == "yes" else 0
    new_review_recommend = np.array([recommend_binary]).reshape(1, -1)
    
    # Combine features
    new_review = hstack([new_review_text, new_review_recommend])
    
    # Make prediction
    prediction = xgb.predict(new_review)
    return int(prediction[0] + 1)

def main(cv, xgb):
    """
    Main function to handle user input and process reviews.
    """
    # Get language preference
    language = input("Is your review in Urdu/Roman Urdu? (yes/no): ").strip().lower()
    is_urdu = language == "yes"
    
    # Get review details
    title = input("Enter your review title: ")
    text = input("Enter your review: ")
    recommend = input("Do you recommend this product? (yes/no): ").strip().lower()
    
    # Process the review
    rating = analyze_review(cv, xgb, title, text, recommend, is_urdu)
    
    # Print results
    print(f"\nReview Analysis Results:")
    print(f"{'Translated ' if is_urdu else ''}Title: {title}")
    print(f"{'Translated ' if is_urdu else ''}Review: {text}")
    print(f"Recommendation: {'Yes' if recommend == 'yes' else 'No'}")
    print(f"Predicted Rating: {rating} stars")

# Usage example:
if __name__ == "__main__":
    # Note: cv and xgb should be your pre-trained CountVectorizer and XGBoost models
    main(cv, xgb)
    pass
```
